# Interpretable Heart Disease Prediction

This notebook documents an independent, reproducible analysis of the processed Cleveland subset of the UCI Heart Disease dataset. The binary target represents absence (0) or presence (1) of heart disease. This educational analysis is not a clinical tool.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from preprocess import load_data, split_data
sns.set_theme(style='whitegrid')
```

## Data loading and quality assessment

```python
data = load_data()
display(data.head())
print(data.shape)
display(data.describe().T)
display(data.isna().sum().rename('missing').to_frame())
```

## Exploratory analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
sns.countplot(data=data, x='target', ax=axes[0])
sns.histplot(data=data, x='age', hue='target', kde=True, ax=axes[1])
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(11, 9))
sns.heatmap(data.corr(numeric_only=True), cmap='vlag', center=0, square=True)
plt.title('Feature Correlation Matrix')
plt.show()
```
